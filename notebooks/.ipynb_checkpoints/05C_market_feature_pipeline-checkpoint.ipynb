# 05C. Market Feature Pipeline

## 목적
05A와 05B에서 수집·정제한 원천 데이터를 이용해 모든 모델이 공통으로 사용할 시장 및 거시경제 Feature를 생성한다.

### Feature Groups
- Macro Features
- Market Features
- Risk Features
- Sector Features

### Point-in-Time 원칙
날짜 t의 종가 및 당일 시장정보로 계산한 Feature는 기본적으로 t+1 거래 의사결정부터 사용할 수 있다고 가정한다.
미래 정보를 이용한 Backfill은 허용하지 않는다.

```python
#05c-2. feature engineering 준비
#프로젝트 경로 설정 및 기준 clean data 불러오기
#입력: 05a에서 저장한 krx300 index data, 05b ecos macro clean data
#주의: krx api 새로 호출 x, 저장되어 있는 로컬 파일만 사용

from pathlib import Path
import numpy as np
import pandas as pd

PROJECT_ROOT = Path(
    r"C:\code\portfolio_optimization"
)

MACRO_PATH = (
    PROJECT_ROOT
    / "data"
    / "clean"
    / "ecos"
    / "macro_daily_clean.csv"
)

KRX300_INDEX_PATH = (
    PROJECT_ROOT
    / "data"
    / "raw"
    / "krx"
    / "index"
    / "krx300_index_daily_pykrx.csv"
)

FEATURE_DIR = (
    PROJECT_ROOT
    / "data"
    / "features"
)

FEATURE_DIR.mkdir(
    parents=True,
    exist_ok=True
)


macro_clean = pd.read_csv(
    MACRO_PATH,
    parse_dates=["date"]
)

print(
    "Macro shape:",
    macro_clean.shape
)

macro_clean.head()
```

```python
#05c-3. 저장된 krx300 index 데이터에서 실제 주식 거래일 추출
#목적: macro 데이터를 한국 주식시장 거래일 기준으로 정렬, 주말/휴장일 제거
#api 호출 x

krx300_index = pd.read_csv(
    KRX300_INDEX_PATH
)

print(
    krx300_index.columns.tolist()
)

krx300_index.head()
```

```python
#trading calendar 정제
#날짜 컬럼을 표준 이름 date로 변경

krx300_index = (
    krx300_index
    .rename(
        columns={
            "날짜": "date"
        }
    )
)

krx300_index["date"] = pd.to_datetime(
    krx300_index["date"]
)

trading_calendar = (
    krx300_index[
        ["date"]
    ]
    .drop_duplicates()
    .sort_values("date")
    .reset_index(drop=True)
)

print(
    "Trading days:",
    len(trading_calendar)
)

print(
    "Start:",
    trading_calendar["date"].min()
)

print(
    "End:",
    trading_calendar["date"].max()
)
```

```python
#05c-4. point-in-time aligment 1
#macro 데이터를 krx 실제 거래일 기준으로 정렬
#방법: krx trading calendar를 기준으로 left join, 해당 거래일에 macro 값이 없으면 과거의 마지막 관측값만 forward fill
#중요: backfill은 사용하지 x, 미래 값을 과거에 채우지 않기 위함임

macro_aligned = (
    trading_calendar
    .merge(
        macro_clean,
        on="date",
        how="left"
    )
    .sort_values("date")
    .reset_index(drop=True)
)

print(
    "Before forward fill:"
)

print(
    macro_aligned.isna().sum()
)
```

```python
#05c-5. forward fill
#point-in-time aligment 2
#거래일에 값이 없는 amcro 변수는 마지막 알려진 값으로 채움

#ffill: forward fill: 과거의 마지막 관측값을 앞으로 전달
#미래값을 과거로 가져오는 bfill은 사용하지 x

MACRO_COLUMNS = [
    "base_rate",
    "usdkrw",
    "bond3y"
]

macro_aligned[
    MACRO_COLUMNS
] = (
    macro_aligned[
        MACRO_COLUMNS
    ]
    .ffill()
)

print(
    "After forward fill:"
)

print(
    macro_aligned.isna().sum()
)

macro_aligned.head()
```

```python
#05c-6. macro feature engineering 1
#거시경제 변수의 일별 변화 feature 생성

#usd/krw_return: 원달러 환율 일간 수익률
#bond3y_change: 국고채 3년 금리 일간 변화폭
#base_rate_change: 한국은행 기준금리 변화폭

macro_features = (
    macro_aligned.copy()
)


macro_features["usdkrw_return_1d"] = (
    macro_features["usdkrw"]
    .pct_change()
)


macro_features["bond3y_change_1d"] = (
    macro_features["bond3y"]
    .diff()
)


macro_features["base_rate_change"] = (
    macro_features["base_rate"]
    .diff()
)

macro_features.head(10)
```

```python
#05c-7. macro feature engineering 2
#단기/중기 macro 변화 feature 생성

#5d: 5 trading days
#20d: 약 1 trading month

macro_features["usdkrw_return_5d"] = (
    macro_features["usdkrw"]
    .pct_change(
        periods=5
    )
)

macro_features["usdkrw_return_20d"] = (
    macro_features["usdkrw"]
    .pct_change(
        periods=20
    )
)


macro_features["bond3y_change_5d"] = (
    macro_features["bond3y"]
    .diff(
        periods=5
    )
)

macro_features["bond3y_change_20d"] = (
    macro_features["bond3y"]
    .diff(
        periods=20
    )
)

macro_features.tail()
```

```python
#05c-8. macro feature engineering 3
#정책금리와 시장금리의 차이 계산

#rate_spread_3y = 국고채 3년 금리 - 한국은행 기준금리
#의미: 정책금리와 시장이 요구하는 중기 금리 사이의 차이

macro_features["rate_spread_3y"] = (
    macro_features["bond3y"]
    - macro_features["base_rate"]
)

macro_features[
    [
        "date",
        "base_rate",
        "bond3y",
        "rate_spread_3y"
    ]
].tail()
```

```python
#05c-9. feature 검증
#macro feature 생성 후 결측치 확인

#주의: 5d/20d rolling feature의 초기 nan은 정상, 미래 데이터로 채우지 xx

print(
    macro_features.isna().sum()
)
```

point-in-time 규칙

모델의 기본 시간 구조:
거래일 t 장 마감 -> t의 가격/환율/금리 데이터 확정 -> feature 계산 -> 모델 판단 -> t+1 거래일에 portfolio 적용

feature(t) -> weight(t+1)

```python
#05c-10. 생성된 macro feature 저장
#목적: macro feature engineering 결과 보존, 이후 market feature와 결합

MACRO_FEATURE_PATH = (
    FEATURE_DIR
    / "macro_features.csv"
)

macro_features.to_csv(
    MACRO_FEATURE_PATH,
    index=False,
    encoding="utf-8-sig"
)

print(
    "Saved:",
    MACRO_FEATURE_PATH
)
```

```python
#05c-11. market feature engineering 준비
#krx300 지수 컬럼명 표준화
#목적: 이후 feature 계산 코드의 일관성 확보

market_data = (
    krx300_index
    .rename(
        columns={
            "시가": "open",
            "고가": "high",
            "저가": "low",
            "종가": "close",
            "거래량": "volume",
            "거래대금": "trading_value",
            "상장시가총액": "market_cap"
        }
    )
    .sort_values("date")
    .reset_index(drop=True)
)

market_data.head()
```

```python
#05c-12. market feature engineering 1
#krx300 시장 수익률 및 momentum 생성
#momentum? 최근 가격 움직임의 방향/강도

#1d:일간, 5d: 약 1주, 20d: 약 1거래월

market_features = (
    market_data.copy()
)

market_features[
    "market_return_1d"
] = (
    market_features["close"]
    .pct_change()
)

market_features[
    "market_return_5d"
] = (
    market_features["close"]
    .pct_change(periods=5)
)

market_features[
    "market_return_20d"
] = (
    market_features["close"]
    .pct_change(periods=20)
)

market_features[
    [
        "date",
        "close",
        "market_return_1d",
        "market_return_5d",
        "market_return_20d"
    ]
].head(25)
```

```python
#05c-13. market feature engineering 2
#krx300 실현 변동성 생성

#realized_volatility=실제 과거 수익률로 계산한 변동성
#20d rolling standard deviation x sqrt(252) -> 연율화 변동성
#np.sqrt(252): 한국 주식시장 통상 약 252거래일 근사 아래 일간 변동성을 연율화하기 위함

market_features[
    "market_volatility_20d"
] = (
    market_features[
        "market_return_1d"
    ]
    .rolling(window=20)
    .std()
    * np.sqrt(252)
)

market_features[
    [
        "date",
        "market_return_1d",
        "market_volatility_20d"
    ]
].head(25)
```

```python
#05c-14. market feature engineering 3
#krx300 drawdown 생성

#drawdown=현재 지수 / 과거 최고 지수 - 1
#ex. 과거 최고 100, 현재 80 -> -20%

market_features[
    "market_peak"
] = (
    market_features["close"]
    .cummax()  #cummax: cumulative maximum: 누적 최대
)

market_features[
    "market_drawdown"
] = (
    market_features["close"]
    / market_features["market_peak"]
    - 1
)

market_features[
    [
        "date",
        "close",
        "market_peak",
        "market_drawdown"
    ]
].tail()
```

```python
#05c-15. market feature engineering4
#krx300 거래량/거래대금 feature 생성
#목적: 시장 거래활동 변화 측정

market_features[
    "volume_change_1d"
] = (
    market_features["volume"]
    .pct_change()
)

market_features[
    "trading_value_change_1d"
] = (
    market_features["trading_value"]
    .pct_change()
)

market_features[
    "trading_value_ma20"
] = (
    market_features["trading_value"]
    .rolling(window=20)
    .mean()
)

market_features[
    [
        "volume_change_1d",
        "trading_value_change_1d",
        "trading_value_ma20"
    ]
].tail()
```

```python
#05c-16. market feature engineering 5
#최근 평균대비 거래대금 수준 계

#trading_value_ratio_20d > 1: 최근 평균보다 거래 활발, < 1: 최근 평균보다 거래 감소

market_features[
    "trading_value_ratio_20d"
] = (
    market_features["trading_value"]
    / market_features["trading_value_ma20"]
)
```

```python
#05c-17. feature 검증
#market feature 생성 결과 및 결측치 확인
#rolling / pct_change 초기 nan은 정상

MARKET_FEATURE_COLUMNS = [
    "market_return_1d",
    "market_return_5d",
    "market_return_20d",
    "market_volatility_20d",
    "market_drawdown",
    "volume_change_1d",
    "trading_value_change_1d",
    "trading_value_ma20",
    "trading_value_ratio_20d"
]

print(
    market_features[
        MARKET_FEATURE_COLUMNS
    ]
    .isna()
    .sum()
)

market_features[
    [
        "date"
    ]
    + MARKET_FEATURE_COLUMNS
].tail()
```

```python
#05c-18. feature 정리
#계산용 중간 컬럼 제거

market_features = (
    market_features
    .drop(
        columns=[
            "market_peak"
        ]
    )
)
```

```python
#05c-19. feature 저장
#krx300 market feature 저장

MARKET_FEATURE_PATH = (
    FEATURE_DIR
    / "krx300_market_features.csv"
)

market_features.to_csv(
    MARKET_FEATURE_PATH,
    index=False,
    encoding="utf-8-sig"
)

print(
    "Saved:",
    MARKET_FEATURE_PATH
)
```

```python
#05c-20. feature 결합
#market feature + macro feature 결합
#기준: 동일한 krx trading date

common_market_features = (
    market_features
    .merge(
        macro_features,
        on="date",
        how="left",
        suffixes=(
            "",
            "_macro"
        )
    )
    .sort_values("date")
    .reset_index(drop=True)
)

print(
    "Shape:",
    common_market_features.shape
)

common_market_features.head()
```

```python
#05c-21. 중간 검증
#market + macro 공통 feature dataset 검증

print(
    "Rows:",
    len(common_market_features)
)

print(
    "Start:",
    common_market_features["date"].min()
)

print(
    "End:",
    common_market_features["date"].max()
)

print(
    "Duplicate dates:",
    common_market_features[
        "date"
    ].duplicated().sum()
)

print(
    "\nMissing values:"
)

print(
    common_market_features
    .isna()
    .sum()
    .sort_values(
        ascending=False
    )
    .head(20)
)
```

```python
#05c-22. feature 정리
#모델 입력 후보 feature만 별도 목록으로 정의

#목적: raw market data 와 derived feature 구분, 이후 ml/rl/regime 모델에서 동일한 feature set 사용

COMMON_FEATURE_COLUMNS = [

    # -------------------------
    # Market Return / Momentum
    # -------------------------
    "market_return_1d",
    "market_return_5d",
    "market_return_20d",

    # -------------------------
    # Market Risk
    # -------------------------
    "market_volatility_20d",
    "market_drawdown",

    # -------------------------
    # Market Activity
    # -------------------------
    "volume_change_1d",
    "trading_value_change_1d",
    "trading_value_ratio_20d",

    # -------------------------
    # Macro Level
    # -------------------------
    "base_rate",
    "usdkrw",
    "bond3y",

    # -------------------------
    # Macro Change
    # -------------------------
    "usdkrw_return_1d",
    "usdkrw_return_5d",
    "usdkrw_return_20d",

    "bond3y_change_1d",
    "bond3y_change_5d",
    "bond3y_change_20d",

    "base_rate_change",

    # -------------------------
    # Rate Environment
    # -------------------------
    "rate_spread_3y"
]


print(
    "Number of common features:",
    len(COMMON_FEATURE_COLUMNS)
)

COMMON_FEATURE_COLUMNS
```

```python
#05c-23. feature dataset 생성
#날짜+모델 입력 feature만 분리
#주의: 아직 shift 하지 x, 아직 target x, 아직 종목별 데이터 x

common_feature_dataset = (
    common_market_features[
        ["date"] + COMMON_FEATURE_COLUMNS
    ]
    .copy()
)

print(
    "Shape:",
    common_feature_dataset.shape
)

common_feature_dataset.head()
```

```python
#05c-24. feature 검증
#rolling feature 계산에 필요한 warm-up 구간 확인
#warm-up: 지표 계산에 필요한 초기 준비 기간
#목적: 실제 모델 사용 가능 시작 날짜 확인, 초기 nan을 미래값으로 채우지 x

feature_ready_mask = (
    common_feature_dataset[
        COMMON_FEATURE_COLUMNS
    ]
    .notna()
    .all(axis=1)
)

print(
    "Total rows:",
    len(common_feature_dataset)
)

print(
    "Fully available rows:",
    feature_ready_mask.sum()
)

print(
    "Warm-up rows:",
    (~feature_ready_mask).sum()
)

print(
    "First fully available date:",
    common_feature_dataset.loc[
        feature_ready_mask,
        "date"
    ].min()
)
```

```python
#05c-25. feature 정제
#모든 공통 feature가 계산 가능한 날짜만 별도 dataset 생성
#목적: 향후 모델 입력용 기본 market context dataset

common_feature_ready = (
    common_feature_dataset[
        feature_ready_mask
    ]
    .reset_index(drop=True)
)

print(
    "Shape:",
    common_feature_ready.shape
)

print(
    "Start:",
    common_feature_ready["date"].min()
)

print(
    "End:",
    common_feature_ready["date"].max()
)

common_feature_ready.head()
```

```python
#05c-26. 최종 저장
#공동 market / macro feature dataset 저장

COMMON_FEATURE_PATH = (
    FEATURE_DIR
    / "common_market_macro_features.csv"
)

COMMON_FEATURE_READY_PATH = (
    FEATURE_DIR
    / "common_market_macro_features_ready.csv"
)


common_feature_dataset.to_csv(
    COMMON_FEATURE_PATH,
    index=False,
    encoding="utf-8-sig"
)

common_feature_ready.to_csv(
    COMMON_FEATURE_READY_PATH,
    index=False,
    encoding="utf-8-sig"
)


print(
    "Full features saved:",
    COMMON_FEATURE_PATH
)

print(
    "Ready features saved:",
    COMMON_FEATURE_READY_PATH
)
```

```python
#05c-27. 현재 feature pipeline 상태 기록

# 완료:
# - ECOS Macro Alignment
# - USD/KRW Features
# - 기준금리 Features
# - 국고채 3년 Features
# - KRX300 Market Return
# - Market Momentum
# - Market Volatility
# - Market Drawdown
# - Volume / Trading Value Features

# 보류:
# - KRX300 Breadth
# - Cross-sectional Dispersion
# - Average Correlation
# - Sector Leadership
# - 종목별 Price / Liquidity Features

# 보류 이유: KRX 개별 종목 / Point-in-Time Universe 데이터 추가 확보 필요

# Point-in-Time 원칙: Feature(t)는 이후 05D에서 t+1 의사결정에 연결

print(
    "05C Market Feature Pipeline:"
    " current available-data stage complete."
)
```
